## 2. The `__init__` constructor

```python
class employee:
    # this is constructor method, it is called automatically when an object is created
    # the first argument is always self, which represents the instance of the class and it is passed automatically when the object is created
    def __init__(self, first_name, last_name, pay):
        # this is same as Employee1.first_name = "John" etc. 
        # self.instance_variable = value(passed here through the constructor arguments)
        self.first_name = first_name
        self.last_name = last_name
        self.email = self.first_name + "." + self.last_name + "@email.com"
        self.pay = pay
```

```python
emp1 = employee("John", "Doe", 50000)
emp2 = employee("Jane", "Doe", 60000)

print(emp1.first_name, emp1.last_name, emp1.email, emp1.pay)
print(emp2.first_name, emp2.last_name, emp2.email, emp2.pay)
```

## 4. Class variables

## 5. Class methods and static methods

| Type | First argument | Use |
|---|---|---|
| instance method | `self` | works with one object |
| `@classmethod` | `cls` | works with the class; alternative constructors |
| `@staticmethod` | none | related utility; uses neither |

```python
# just like instance variable and class variables
# we have instance methods, class methods, and static methods

# how can a method take the class as its first argument instead of the instance?
# we use the @classmethod decorator and pass cls as the first argument
class Employee:
    num_of_emps = 0
    raise_amount = 1.06

    def __init__(self, first_name, last_name, email, pay):
        self.first_name = first_name
        self.last_name = last_name
        self.email = email
        self.pay = pay
        Employee.num_of_emps += 1

    def full_name(self):
        return f"{self.first_name} {self.last_name}"

    def apply_raise(self):
        self.pay = int(self.pay * self.raise_amount)

    @classmethod
    def set_raise_amount(cls, amount):
        cls.raise_amount = amount

emp1 = Employee("John", "Doe", "john.doe@example.com", 50000)
emp2 = Employee("Jane", "Smith", "jane.smith@example.com", 60000)

print(emp1.raise_amount)
print(emp2.raise_amount)
print(Employee.raise_amount)
```

### Static methods

```python
import datetime

class Employee:
    num_of_emps = 0
    raise_amount = 1.06

    def __init__(self, first_name, last_name, pay):
        self.first_name = first_name
        self.last_name = last_name
        self.email = f"{self.first_name}.{self.last_name}@example.com"
        self.pay = pay
        Employee.num_of_emps += 1

    def full_name(self):
        return f"{self.first_name} {self.last_name}"

    def apply_raise(self):
        self.pay = int(self.pay * self.raise_amount)

    @classmethod
    def set_raise_amount(cls, amount):
        cls.raise_amount = amount

    @classmethod
    def from_string(cls, emp_str): #pass a string in the format "first-last-pay"
        first, last, pay = emp_str.split("-")
        return cls(first, last, int(pay))

    @staticmethod
    def is_workday(day):  # no self or cls: it does not use the instance or the class
        # weekday(): Monday=0 ... Saturday=5, Sunday=6
        if day.weekday() == 5 or day.weekday() == 6:
            return False
        return True


emp1 = Employee("John", "Doe", 50000)
emp2 = Employee("Jane", "Smith", 60000)


my_date = datetime.date(2001, 4, 23) #create a date object
print(Employee.is_workday(my_date))
```

### Overriding `__init__` with `super()`

```python
class Employee:

    raise_amount = 1.06

    def __init__(self, first_name, last_name, pay):
        self.first_name = first_name
        self.last_name = last_name
        self.email = f"{self.first_name}.{self.last_name}@example.com"
        self.pay = pay

    def full_name(self):
        return f"{self.first_name} {self.last_name}"

    def apply_raise(self):
        self.pay = int(self.pay * self.raise_amount)

class Developer(Employee):
    raise_amount = 1.10
    def __init__(self, first_name, last_name, pay, prog_lang):
        super().__init__(first_name, last_name, pay)
        self.prog_lang = prog_lang


dev1 = Developer("John", "Doe", 50000, "Python")
dev2 = Developer("Jane", "Smith", 60000, "JavaScript")

print(dev1.full_name(), dev1.prog_lang)
print(dev2.full_name(), dev2.prog_lang)

# Developer now overrides __init__ to add prog_lang.
# super().__init__(...) lets the parent (Employee) set first_name, last_name, email, pay,
# so we don't repeat that code. Developer also overrides the class variable raise_amount.
```

```python
class Employee:

    raise_amount = 1.06

    def __init__(self, first_name, last_name, pay):
        self.first_name = first_name
        self.last_name = last_name
        self.email = f"{self.first_name}.{self.last_name}@example.com"
        self.pay = pay

    def full_name(self):
        return f"{self.first_name} {self.last_name}"

    def apply_raise(self):
        self.pay = int(self.pay * self.raise_amount)

class Developer(Employee):
    raise_amount = 1.10
    def __init__(self, first_name, last_name, pay, prog_lang):
        super().__init__(first_name, last_name, pay)
        self.prog_lang = prog_lang

class Manager(Employee):
    # employees=None, not employees=[]: a mutable default list would be shared by every Manager
    def __init__(self, first_name, last_name, pay, employees=None):
        super().__init__(first_name, last_name, pay)
        self.employees = employees if employees is not None else []

    def add_employee(self, emp):
        if emp not in self.employees:
            self.employees.append(emp)

    def remove_employee(self, emp):
        if emp in self.employees:
            self.employees.remove(emp)

    def print_emps(self):
        for emp in self.employees:
            print("-->", emp.full_name())

dev1 = Developer("John", "Doe", 50000, "Python")
dev2 = Developer("Jane", "Smith", 60000, "JavaScript")

mgr1 = Manager("Alice", "Johnson", 80000, [dev1])
# print(mgr1.email)

# mgr1.print_emps()
mgr1.add_employee(dev2)
mgr1.remove_employee(dev1)
mgr1.print_emps()
```

### `isinstance` and `issubclass`

```python
class Employee:

    raise_amount = 1.06

    def __init__(self, first_name, last_name, pay):
        self.first_name = first_name
        self.last_name = last_name
        self.email = f"{self.first_name}.{self.last_name}@example.com"
        self.pay = pay

    def full_name(self):
        return f"{self.first_name} {self.last_name}"

    def apply_raise(self):
        self.pay = int(self.pay * self.raise_amount)

class Developer(Employee):
    raise_amount = 1.10
    def __init__(self, first_name, last_name, pay, prog_lang):
        super().__init__(first_name, last_name, pay)
        self.prog_lang = prog_lang

class Manager(Employee):
    def __init__(self, first_name, last_name, pay, employees=None):
        super().__init__(first_name, last_name, pay)
        self.employees = employees if employees is not None else []

    def add_employee(self, emp):
        if emp not in self.employees:
            self.employees.append(emp)

    def remove_employee(self, emp):
        if emp in self.employees:
            self.employees.remove(emp)

    def print_emps(self):
        for emp in self.employees:
            print("-->", emp.full_name())

dev1 = Developer("John", "Doe", 50000, "Python")
dev2 = Developer("Jane", "Smith", 60000, "JavaScript")

mgr1 = Manager("Alice", "Johnson", 80000, [dev1])
print(isinstance(dev1, Developer))
print(isinstance(dev1, Employee))  # True: a Developer is also an Employee
print(isinstance(dev1, Manager))
```

```python
print(issubclass(Developer, Employee))
print(issubclass(Manager, Employee))
print(issubclass(Developer, Manager))
print(issubclass(Manager, Developer))
```

## 7. Special (dunder) methods

```python
class Employee:

    raise_amount = 1.06

    def __init__(self, first_name, last_name, pay):
        self.first_name = first_name
        self.last_name = last_name
        self.email = f"{self.first_name}.{self.last_name}@example.com"
        self.pay = pay

    def full_name(self):
        return f"{self.first_name} {self.last_name}"

    def apply_raise(self):
        self.pay = int(self.pay * self.raise_amount)

    def __repr__(self):
        return f"Employee('{self.first_name}', '{self.last_name}', {self.pay})"
    def __str__(self):
        return f"{self.full_name()} <{self.email}>"

    def __add__(self, other):
        return self.pay + other.pay

    def __len__(self):
        return len(self.full_name())
    
    

emp1 = Employee("John", "Doe", 50000)
emp2 = Employee("Jane", "Smith", 60000)

print(emp1)  # print() uses __str__ if defined, otherwise falls back to __repr__
print(emp2)

# to get repr, use repr() function
print(repr(emp1))
print(repr(emp2))

print(emp1 + emp2)  # calls emp1.__add__(emp2)
print(len(emp1))  # calls emp1.__len__()
```

## 8. Property decorators: getter, setter, deleter

```python
class employee:
    raise_amount = 1.06

    def __init__(self, first_name, last_name, pay):
        self.first_name = first_name
        self.last_name = last_name
        self.email = f"{self.first_name}.{self.last_name}@example.com"
        self.pay = pay

    def full_name(self):
        return f"{self.first_name} {self.last_name}"

emp1 = employee("John", "Doe", 50000)

emp1.first_name = "jake"

print(emp1.first_name)
print(emp1.email) # email is not updated after changing first_name
print(emp1.pay)
print(emp1.full_name()) # full name is updated after changing first_name
```

```python
# How do we update email automatically when first_name changes?
# @property: define a method, but access it like an attribute (no parentheses)

class employee:
    raise_amount = 1.06

    def __init__(self, first_name, last_name, pay):
        self.first_name = first_name
        self.last_name = last_name
        self.pay = pay

    @property
    def email(self):
        return f"{self.first_name}.{self.last_name}@example.com"
    @property
    def full_name(self):
        return f"{self.first_name} {self.last_name}"

    @full_name.setter
    def full_name(self, value):
        self.first_name, self.last_name = value.split()

    @full_name.deleter
    def full_name(self):
        print("Deleting full name")
        self.first_name = None
        self.last_name = None

emp1 = employee("John", "Doe", 50000)
emp1.first_name = "jake"
print(emp1.first_name)
print(emp1.email) # email is automatically updated after changing first_name, even though it is not stored as an attribute
print(emp1.pay)
print(emp1.full_name) # full name is updated after changing first_name

emp1.full_name = "pank smith"
print(emp1.first_name)
print(emp1.last_name)
print(emp1.email) # email is automatically updated after changing full_name
print(emp1.full_name) # full name is updated after changing full_name

del emp1.full_name
print(emp1.first_name)
print(emp1.last_name)
print(emp1.email) # email is automatically updated after deleting full_name
print(emp1.full_name) # full name is updated after deleting full_name
```
